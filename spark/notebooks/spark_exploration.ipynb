# Spark exploration notebook

This notebook demonstrates how to create a local Spark session, read Parquet files,
and inspect schema inference. Use this for local development and exploratory testing.

```python
# Build a local SparkSession for exploration
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName('spark_exploration')
    .master('local[*]')
    .config('spark.sql.session.timeZone', 'UTC')
    .getOrCreate()
)
spark
```

```python
# Read Parquet and show schema (replace the path with your local or gs:// path)
path = '/tmp/btc_trades'  # example local path or gs://bucket/path
df = spark.read.option('mergeSchema', 'true').parquet(path)
df.printSchema()
df.show(5, truncate=False)
```

```python
# Compare read with and without mergeSchema
df1 = spark.read.option('mergeSchema', 'false').parquet(path)
df2 = spark.read.option('mergeSchema', 'true').parquet(path)
print('schema (no merge):')
df1.printSchema()
print('schema (merge):')
df2.printSchema()
```
